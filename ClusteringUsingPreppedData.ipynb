## Select relevant features

### Subtask:
Choose the columns from the merged DataFrame that will be used for clustering. It's important to select features that are relevant to the type of grouping you want to achieve.


**Reasoning**:
Examine the columns of the merged_df to identify features relevant for clustering and select a subset of numerical columns for clustering, excluding identifiers, dates, and other non-numeric or irrelevant columns.



```python
# Select numerical features for clustering from merged_df
# Exclude identifier columns, date/time columns, and other non-numeric columns
features = merged_df.select_dtypes(include=['int64', 'float64']).columns.tolist()

# Remove specific columns that are not suitable for clustering
exclude_cols = [
    'OBJECTID',
    'REPORT_YEAR', 'REPORT_DAY', 'REPORT_DOY', 'REPORT_HOUR',
    'OCC_YEAR', 'OCC_DAY', 'OCC_DOY', 'OCC_HOUR',
    'UCR_CODE', 'UCR_EXT',
    'HOOD_158', 'HOOD_140', 'Neighbourhood_Number',
    'REPORT_MONTH_SEQ', 'DAY_SEQ', 'WEEK_SEQ',
    'x', 'y', 'LONG_WGS84', 'LAT_WGS84' # Exclude location coordinates for now
]

clustering_features = [col for col in features if col not in exclude_cols]

# Display the selected features
print("Selected features for clustering:")
print(clustering_features)

# Create a new DataFrame with the selected features
X = merged_df[clustering_features]

# Display the first few rows of the feature DataFrame
display(X.head())
```

```python
from sklearn.impute import SimpleImputer

# Handle missing values by imputing with the mean
imputer = SimpleImputer(strategy='mean')
X_processed = imputer.fit_transform(X)

# Create a new DataFrame with processed features (optional, but good for inspection)
X_processed_df = pd.DataFrame(X_processed, columns=clustering_features)

print("Processed features (missing values imputed):")
display(X_processed_df.head())
```

## Choose and Apply Clustering Models

### Subtask:
Apply K-Means and DBSCAN clustering models to the preprocessed data.

**Reasoning**:
Apply two different clustering algorithms: K-Means and DBSCAN. For K-Means, we will choose an arbitrary number of clusters (e.g., 5) for demonstration purposes. For DBSCAN, we will use default parameters. The results of the clustering will be added as new columns to the `merged_df` DataFrame.

```python
from sklearn.cluster import KMeans, DBSCAN

# Apply K-Means clustering
# We need to choose a number of clusters (n_clusters). Let's start with 5 as an example.
kmeans = KMeans(n_clusters=5, random_state=42, n_init=10) # Added n_init to suppress warning
merged_df['kmeans_cluster'] = kmeans.fit_predict(X_processed)

# Apply DBSCAN clustering
# DBSCAN does not require specifying the number of clusters beforehand.
# We will use default values for eps and min_samples for now.
dbscan = DBSCAN()
merged_df['dbscan_cluster'] = dbscan.fit_predict(X_processed)

# Display the first few rows of the merged DataFrame with cluster labels
display(merged_df.head())
```

## Evaluate clustering results

### Subtask:
Evaluate the results of the K-Means and DBSCAN clustering models.

**Reasoning**:
Evaluate the clustering results by examining the number of data points in each cluster for both K-Means and DBSCAN. This will give us an initial understanding of how the data has been grouped by each algorithm.

```python
# Evaluate K-Means clustering results
print("K-Means Cluster Distribution:")
print(merged_df['kmeans_cluster'].value_counts())

# Evaluate DBSCAN clustering results
print("\nDBSCAN Cluster Distribution:")
print(merged_df['dbscan_cluster'].value_counts())
```

## Experiment with DBSCAN Parameters

### Subtask:
Apply DBSCAN with a few different combinations of `eps` and `min_samples` to observe the effect on clustering.

**Reasoning**:
Apply DBSCAN with 3-4 different combinations of `eps` and `min_samples` to demonstrate how the number of clusters and noise points change. We will use arbitrary values for demonstration purposes, as systematic tuning requires more involved methods. The results will be added as new columns to the `merged_df` DataFrame.

```python
from sklearn.cluster import DBSCAN

# Experiment with different DBSCAN parameters

# Parameter combination 1: Slightly larger eps
dbscan_exp1 = DBSCAN(eps=0.5, min_samples=5) # Default eps is 0.5, let's try 0.7
merged_df['dbscan_cluster_exp1'] = dbscan_exp1.fit_predict(X_processed)
print("\nDBSCAN Experiment 1 (eps=0.5, min_samples=5) Cluster Distribution:")
print(merged_df['dbscan_cluster_exp1'].value_counts())


# Parameter combination 2: Larger eps
dbscan_exp2 = DBSCAN(eps=1.0, min_samples=5)
merged_df['dbscan_cluster_exp2'] = dbscan_exp2.fit_predict(X_processed)
print("\nDBSCAN Experiment 2 (eps=1.0, min_samples=5) Cluster Distribution:")
print(merged_df['dbscan_cluster_exp2'].value_counts())


# Parameter combination 3: Larger min_samples
dbscan_exp3 = DBSCAN(eps=0.5, min_samples=10)
merged_df['dbscan_cluster_exp3'] = dbscan_exp3.fit_predict(X_processed)
print("\nDBSCAN Experiment 3 (eps=0.5, min_samples=10) Cluster Distribution:")
print(merged_df['dbscan_cluster_exp3'].value_counts())

# Parameter combination 4: Larger eps and larger min_samples
dbscan_exp4 = DBSCAN(eps=1.0, min_samples=10)
merged_df['dbscan_cluster_exp4'] = dbscan_exp4.fit_predict(X_processed)
print("\nDBSCAN Experiment 4 (eps=1.0, min_samples=10) Cluster Distribution:")
print(merged_df['dbscan_cluster_exp4'].value_counts())

# Display the first few rows of the merged DataFrame with new cluster labels
display(merged_df.head())
```

## Discuss clustering evaluation metrics

### Subtask:
Explain different metrics used to evaluate clustering results, including intrinsic metrics like Silhouette Score, Davies-Bouldin Index, and Calinski-Harabasz Index.


**Reasoning**:
Provide explanations of the clustering evaluation metrics as requested by the subtask.



```python
print("--- Clustering Evaluation Metrics ---")

print("\nImportance of Clustering Evaluation:")
print("Clustering is an unsupervised learning technique, meaning we don't have ground truth labels to directly assess the accuracy of the clusters. Clustering evaluation metrics provide ways to quantify the quality of the clustering by assessing how well-separated and compact the clusters are. This helps in comparing different clustering algorithms or different parameter settings for the same algorithm.")

print("\n1. Silhouette Score:")
print("The Silhouette Score measures how similar an object is to its own cluster compared to other clusters. It ranges from -1 to +1.")
print("- A score near +1 indicates that the data point is well within its own cluster and far from other clusters (good clustering).")
print("- A score near 0 indicates that the data point is close to the decision boundary between two clusters.")
print("- A score near -1 indicates that the data point might have been assigned to the wrong cluster.")
print("The average Silhouette Score across all data points gives an overall measure of clustering quality.")
print("It is calculated for each sample using the mean intra-cluster distance (a) and the mean nearest-cluster distance (b): (b - a) / max(a, b).")

print("\n2. Davies-Bouldin Index:")
print("The Davies-Bouldin Index measures the average similarity ratio of each cluster with its most similar cluster. Similarity is measured by the ratio of within-cluster scatter to between-cluster separation.")
print("- A lower Davies-Bouldin Index indicates better clustering, meaning clusters are more separated and less scattered internally.")
print("- The minimum possible value is 0.")

print("\n3. Calinski-Harabasz Index:")
print("The Calinski-Harabasz Index (also known as the Variance Ratio Criterion) measures the ratio of the between-cluster variance to the within-cluster variance.")
print("- A higher Calinski-Harabasz Index generally indicates better clustering, with well-defined clusters.")
print("- It is calculated as the ratio of the sum of between-cluster dispersion and within-cluster dispersion.")

print("\nIntrinsic Metrics:")
print("Silhouette Score, Davies-Bouldin Index, and Calinski-Harabasz Index are all intrinsic evaluation metrics. This means they assess the quality of a clustering based solely on the dataset and the resulting clustering itself. They do not require external information or ground truth labels to evaluate the clustering.")
```
